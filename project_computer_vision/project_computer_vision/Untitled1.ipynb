```python
import os
import shutil
from sklearn.model_selection import train_test_split

dataset_path = "d2"

classes = [
    "destroyed",
    "damaged",
    "no_damage",
    "Missile-Damaged Building"
]

for split in ["train", "validation", "test"]:
    for cls in classes:
        os.makedirs(os.path.join(dataset_path, split, cls), exist_ok=True)

for cls in classes:

    source_folder = os.path.join(dataset_path, cls)

    images = [
        f for f in os.listdir(source_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    train_imgs, temp_imgs = train_test_split(
        images,
        test_size=0.30,
        random_state=42
    )

    val_imgs, test_imgs = train_test_split(
        temp_imgs,
        test_size=0.50,
        random_state=42
    )

    for img in train_imgs:
        shutil.copy(
            os.path.join(source_folder, img),
            os.path.join(dataset_path, "train", cls, img)
        )

    for img in val_imgs:
        shutil.copy(
            os.path.join(source_folder, img),
            os.path.join(dataset_path, "validation", cls, img)
        )

    for img in test_imgs:
        shutil.copy(
            os.path.join(source_folder, img),
            os.path.join(dataset_path, "test", cls, img)
        )

print("Dataset split completed successfully!")
```
